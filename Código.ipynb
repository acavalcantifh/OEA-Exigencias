```python
#Bibliotecas
import pandas as pd
import os
os.chdir('/content/drive/My Drive/Colab Notebooks/PósPUC - TCC/Base testes')
ds=pd.read_excel('Exigencias despacho – Base testes.xlsx')

#Camuflar nomes das empresas importadoras
ImpUniques= ds["Nome"].unique()
k=0
for i in ImpUniques:
    k=k+1
    for index, row in ds.iterrows():
        if i == row["Nome"]:
            ds.loc[index,"Nome"]="Imp"+str(k)

ds.to_excel(r'ds_camuflado.xlsx',index= False)
```
